## Extraire les mots clés d'un document avec Yake

https://github.com/INESCTEC/yake

```python
# Instantier l'extracteur de mots clés
kw_extractor = yake.KeywordExtractor(lan="fr", top=50)
kw_extractor
```

```python
# Lister les Fichiers
data_path = "../../data/txt/"
files = [f for f in os.listdir(data_path) if f.endswith('.txt')]
```

```python
# Imprimer le nombre de fichiers identifiés
len(files)
```

# choix de l’année et filtre

```python
annee = "1964"  # l'année choisie

#voir  tous les fichiers du dossier
files = os.listdir(data_path)

# on a une liste vide, et on y ajoute les fichiers de 1964
files_annee = []
for f in files:
    # on garde seulement les fichiers .txt dont le nom contient l'année
    if f.endswith(".txt") and annee in f:
        files_annee.append(f)

print(len(files_annee), "fichiers pour", annee)  # doit afficher 100
```

```python
# Choisir un fichier
this_file = files[0]
this_file
```

```python
# Récupérer le texte du fichier
text = open(os.path.join(data_path, this_file), 'r', encoding='utf-8').read()
text[:500]
```

```python
# Extraire les mots clés de ce texte
keywords = kw_extractor.extract_keywords(text)
keywords
```

```python
# Ne garder que les bigrammes
kept = []
for kw, score in keywords:
    words = kw.split()
    if len(words) == 2:
        kept.append(kw)
kept
```

## Faire la même opération sur tous les documents

```python
for f in sorted(files)[:10]:
    text = open(os.path.join(data_path, f), 'r', encoding="utf-8").read()
    keywords = kw_extractor.extract_keywords(text)
    kept = []
    for kw, score in keywords:
        words = kw.split()
        if len(words) == 2:
            kept.append(kw)
    print(f"{f} mentions these keywords: {', '.join(kept)}...")
```
